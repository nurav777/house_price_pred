```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Housing-selected-columns.csv")
print(df.shape)
```

```python
categorical_cols=['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning']
df[categorical_cols] = df[categorical_cols].replace({"yes":1,"no":0})
```

Convert it to train_data.csv and test_data.csv




```python
from sklearn.model_selection import train_test_split
train_data,test_data=train_test_split(df,test_size=0.3,random_state=42)
train_data.to_csv("train_data.csv",index=False)
test_data.to_csv("test_data.csv",index=False)
```
